1: Calculate the number of digits and letters in a string | 15 points
Write a program that accepts a string and calculates the number of digits and letters

Example:
Input: Benjamin123
Output:
    Digits: 3
    Letters: 8


```python
word = input("Enter a string:")
digit = 0
letters = 0

for letter in word:
    if letter.isdigit():
        digit += 1
    elif letter.isalpha():
        letters += 1
    else:
        pass
        
print("Digits:", digit)
print("Letters:", letters)
```

2: Masking a Credit Card Number | 15 points
Write a program that takes a credit card number as input, checks the length (16) to know if it's a valid card or not, and then hides all but the last four digits..

Example: 
Input: 1234567890123456
Output: ************3456



```python
card_number = input("Enter your credit card number: ")
if len(card_number) == 16 and card_number.isdigit():
    masked_number = "*" * 12 + card_number[-4:]
    print(masked_number)
else:
    print("Invalid card number")
```

Ask the user to enter a post sentence. Then, check for any word from this list sensitive words ["bad", "ugly", "stupid", 'hate', 'sexy'] in the sentence. Replace each of those words with **** and print the cleaned sentence.

3: AI Training Data Cleaner | 10 points
Skills tested: List comprehensions/for loops, Strings, String Methods, Functions, Data Cleaning Logic
You're preparing a raw text dataset for training an NLP model on customer reviews. The raw dataset contains empty strings, whitespace-only entries, and very short entries (fewer than 3 words) that add noise without useful signal. You need to filter the dataset down to only the clean, usable reviews before tokenization. Note: raw reviews is a list of strings (may include empty strings or whitespace).

Example input:
    raw_reviews = [
    "This product is amazing and works well",
    "",
    "  "
    "Good",
    "Not bad at all",
    "Excellent service!",
]

Output:
    ['This product is amazing and works well', 'Not bad at all']

Write a function 'clean_training_reviews' that strip whitespace from each review, split each review into words to count them, keep only reviews where the word count is 3 or more, then returns a list of the cleaned data. ⁠

```python
raw_data =  "This product is amazing and works well", "", " " "Good", "Not bad at all", "Excellent service!" 
def clean_training_reviews(raw_data):
    output =  []
    for data in raw_data:
        if len(data.strip().split()) > 3:
            output.append(data)
    return output

clean_training_reviews(raw_data)
```
